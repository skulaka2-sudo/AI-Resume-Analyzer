```python
import pandas as pd

job_roles = {
    "Job Role": [
        "Data Analyst",
        "Machine Learning Engineer",
        "AI Engineer",
        "NLP Engineer",
        "Computer Vision Engineer"
    ],
    "Skills": [
        "Python, SQL, Excel, Pandas, Power BI",
        "Python, Machine Learning, Scikit-learn, FastAPI, Docker",
        "Python, Deep Learning, LLM, RAG, APIs",
        "Python, NLP, Transformers, Hugging Face",
        "Python, OpenCV, CNN, YOLO"
    ]
}

job_df = pd.DataFrame(job_roles)

job_df.to_csv("job_roles.csv", index=False)

print(job_df)
```

```python
import pandas as pd

skills = [
    "Python",
    "Java",
    "C",
    "C++",
    "SQL",
    "Excel",
    "Power BI",
    "Pandas",
    "NumPy",
    "Machine Learning",
    "Deep Learning",
    "NLP",
    "TensorFlow",
    "PyTorch",
    "Scikit-learn",
    "Docker",
    "FastAPI",
    "OpenCV",
    "CNN",
    "YOLO",
    "Transformers",
    "Hugging Face",
    "Git",
    "GitHub",
    "LLM",
    "RAG",
    "APIs",
    "Flask",
    "Streamlit",
    "SQLite"
]

skill_df = pd.DataFrame(skills, columns=["Skill"])

skill_df.to_csv("skill_dictionary.csv", index=False)

print(skill_df)
```

```python
sample_resume = """
John Smith

Email: johnsmith@gmail.com
Phone: +1 9876543210

Education
Bachelor of Technology in Computer Science

Skills
Python
SQL
Pandas
NumPy
Machine Learning
Scikit-learn
Git
GitHub
Excel

Projects
Student Performance Prediction using Machine Learning
Weather Prediction System
AI Chatbot using Python

Experience
Completed AI & ML Internship.
Worked on data analysis and machine learning projects.
"""

with open("sample_resume.txt", "w") as file:
    file.write(sample_resume)

print("Sample resume created successfully!")
```

```python
with open("sample_resume.txt", "r") as file:
    resume_text = file.read()

print(resume_text)
```

```python
import re

def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove special characters except +
    text = re.sub(r'[^a-zA-Z0-9+\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()

cleaned_resume = clean_text(resume_text)

print(cleaned_resume)
```

```python
import pandas as pd

skills_df = pd.read_csv("skill_dictionary.csv")

skill_list = skills_df["Skill"].tolist()

print(skill_list)
```

```python
found_skills = []

for skill in skill_list:
    if skill.lower() in cleaned_resume:
        found_skills.append(skill)

print("Skills Found:")
print(found_skills)
```

```python
import pandas as pd

# Load the job roles dataset
job_df = pd.read_csv("job_roles.csv")

print(job_df)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Combine resume text and job role skills
documents = [cleaned_resume] + job_df["Skills"].tolist()

print(documents)
```

```python
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print(tfidf_matrix.shape)
```

```python
similarity_scores = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:]
)

print(similarity_scores)
```

```python
job_df["Match Score"] = similarity_scores[0] * 100

job_df[["Job Role", "Match Score"]]
```

```python
top_jobs = job_df.sort_values(by="Match Score", ascending=False)

print("Top 3 Recommended Jobs:\n")

top_jobs[["Job Role", "Match Score"]].head(3)
```

```python
# Select the best matching job
best_job = top_jobs.iloc[0]

print("Best Job Role:", best_job["Job Role"])

required_skills = [skill.strip() for skill in best_job["Skills"].split(",")]

print("\nRequired Skills:")
print(required_skills)

print("\nYour Skills:")
print(found_skills)

missing_skills = []

for skill in required_skills:
    if skill not in found_skills:
        missing_skills.append(skill)

print("\nMissing Skills:")
print(missing_skills)
```

```python
print("\n===== LEARNING ROADMAP =====")

if len(missing_skills) == 0:
    print("Congratulations! Your resume already contains all the required skills.")
else:
    for i, skill in enumerate(missing_skills, start=1):
        print(f"Week {i}: Learn {skill}")
```
